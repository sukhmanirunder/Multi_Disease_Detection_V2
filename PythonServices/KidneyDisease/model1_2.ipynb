```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
from sklearn.model_selection import GridSearchCV
```

```python
data=pd.read_csv('kidney3.csv')
data
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
print(data.duplicated().sum())
```

```python
x= data.drop('Chronic Kidney Disease: yes', axis=1)
y = data['Chronic Kidney Disease: yes']
```

```python
print(y.value_counts())
```

```python
corr = data.corr()

corr["Chronic Kidney Disease: yes"].sort_values(ascending=False)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
x_train.shape,x_test.shape
```

```python
ss=StandardScaler()
x_train_ss=ss.fit_transform(x_train)
x_test_ss=ss.fit_transform(x_test)
x_train_ss
```

```python
x_train_ss=pd.DataFrame(x_train_ss,columns=x_train.columns)
x_test_ss=pd.DataFrame(x_test_ss,columns=x_test.columns)
x_train_ss
```

```python
param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [3, 5, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
```

```python
rf=RandomForestClassifier(n_estimators=100,random_state=45)
```

```python
grid = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=2
)
```

```python
grid.fit(x_train_ss, y_train)
```

```python
print(grid.best_params_)
```

```python
best_model = grid.best_estimator_
```

```python
y_pred = best_model.predict(x_test_ss)
print(y_pred)
```

```python
a=accuracy_score(y_pred,y_test)
a
```

```python
print(confusion_matrix(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```
